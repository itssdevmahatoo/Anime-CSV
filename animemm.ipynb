```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("anime.csv")
```

```python
df = pd.DataFrame(df)
```

```python
df
```

**Make a new column for episode count**

```python
def extract_episodes(txt):
    check = False 
    data = ""
    for i in txt:
        if i == ")" :
            check = False
            return data
        if check == True:
            data = data + i
        if i == "(":
            check = True
```

```python
df["Episodes"] = df["Title"].apply(extract_episodes)
```

```python
df
```

```python
df["Episodes"] = df["Episodes"].str.replace("eps","")
```

```python
df
```

```python
df['Episodes'] = df['Episodes'].astype(int)
```

```python
df
```

***Make a new column for time stamp***

```python
def extraction_time(txt):
    check = False
    data = " "
    for i in range(len(txt)):
        if txt[i] == ')':
            for j in range(i+1,i+20):
                data += txt[j]

            return data
```

```python
df["Time Duration"] = df["Title"].apply(extraction_time)
```

```python
df
```

***Give me top 5 highest scoring anime***

```python
df.sort_values("Score", ascending = False)[["Rank","Title","Score"]].head(5)
```

```python
Top=df.sort_values("Score", ascending = False)[["Rank","Title","Score"]].head(5)

plt.scatter(Top["Rank"], Top["Score"])
plt.xlabel("Rank")
plt.ylabel("Score")
plt.title("Anime vs Score")
plt.xticks(rotation=0)
plt.show()
```
